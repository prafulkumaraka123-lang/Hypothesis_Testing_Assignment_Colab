## Question 1 — Hypothesis testing and key terms

**Hypothesis testing** is a statistical procedure that uses sample data to evaluate a claim about a population. It helps determine whether the observed sample evidence is sufficiently inconsistent with a stated claim to reject it.

- **Null hypothesis (H₀):** The default claim, usually representing no effect, no difference, or a specified population value.
- **Alternative hypothesis (H₁):** The claim supported when the null hypothesis is rejected.
- **Significance level (α):** The chosen maximum probability of a Type I error. A common value is 0.05.
- **Test statistic:** A value calculated from sample data (such as z, t, or chi-square) and used to assess the evidence against H₀.
- **p-value:** Assuming H₀ is true, the probability of obtaining a test statistic at least as extreme as the observed one, in the direction(s) specified by H₁.
- **Decision rule:** Reject H₀ if p-value ≤ α; otherwise, fail to reject H₀.

**One-tailed test:** Tests for an effect in one specified direction. Example: H₀: μ = 50 versus H₁: μ > 50.

**Two-tailed test:** Tests for a difference in either direction. Example: H₀: μ = 50 versus H₁: μ ≠ 50.

## Question 2 — Type I error, Type II error and power

Consider a medical test where H₀ says a patient does not have a disease and H₁ says the patient has the disease.

- **Type I error:** Rejecting a true H₀. In this example, concluding that a healthy patient has the disease (false positive).
- **Type II error:** Failing to reject a false H₀. In this example, concluding there is insufficient evidence of disease when the patient is actually sick (false negative).

The probability of a Type I error is α, when the test is calibrated to the chosen significance level. The **power** of a test is the probability of correctly rejecting a false null hypothesis: \(1-\beta\), where β is the probability of a Type II error.

Power can be increased by using a larger sample, reducing measurement variability, using a more powerful appropriate design/test, or (with the trade-off of a higher Type I error risk) increasing α.

| Reality | Fail to reject H₀ | Reject H₀ |
|---|---|---|
| H₀ is true | Correct decision | Type I error |
| H₀ is false | Type II error | Correct decision (power) |

## Question 3 — One-sample z-test for bag weights

Given population standard deviation σ = 8 kg, sample size n = 64, sample mean = 52 kg, and hypothesized mean μ₀ = 50 kg.

**Hypotheses (two-tailed):**
- H₀: μ = 50 kg
- H₁: μ ≠ 50 kg

Standard error:

\[
SE=\frac{\sigma}{\sqrt{n}}=\frac{8}{\sqrt{64}}=1
\]

Test statistic:

\[
z=\frac{\bar{x}-\mu_0}{SE}=\frac{52-50}{1}=2
\]

At α = 0.05 for a two-tailed test, the critical values are approximately −1.96 and +1.96. The two-tailed p-value for z = 2 is approximately 0.0455.

Since 2 > 1.96 and 0.0455 < 0.05, reject H₀. There is statistically significant evidence at the 5% level that the population mean bag weight differs from 50 kg.

```python
import numpy as np
from scipy import stats

mu0, sigma, n, xbar = 50, 8, 64, 52
alpha = 0.05
se = sigma / np.sqrt(n)
z_stat = (xbar - mu0) / se
p_value = 2 * stats.norm.sf(abs(z_stat))
critical = stats.norm.ppf(1 - alpha / 2)

print("Standard error:", se)
print("z statistic:", z_stat)
print("Critical values:", -critical, critical)
print("Two-tailed p-value:", p_value)
print("Decision:", "Reject H0" if p_value < alpha else "Fail to reject H0")
```

## Question 4 — One-proportion z-test

Given n = 200, 120 satisfied customers, and claimed proportion p₀ = 0.5.

\[
\hat p=\frac{120}{200}=0.60
\]

Hypotheses: H₀: p = 0.50; H₁: p ≠ 0.50.

Standard error under H₀:

\[
SE=\sqrt{\frac{p_0(1-p_0)}{n}}
=\sqrt{\frac{0.5(0.5)}{200}}\approx0.03536
\]

Test statistic:

\[
z=\frac{0.60-0.50}{0.03536}\approx2.8284
\]

The two-tailed p-value is approximately 0.0047. Since p-value < 0.05, reject H₀: the data provide evidence that the satisfaction proportion differs from 0.5.

A 95% **Wald confidence interval** is:

\[
\hat p\pm1.96\sqrt{\frac{\hat p(1-\hat p)}{n}}
=0.60\pm1.96\sqrt{\frac{0.6(0.4)}{200}}
\]

This gives approximately (0.532, 0.668). Since 0.50 is outside this interval, it supports rejecting H₀.

```python
import numpy as np
from scipy import stats

n, successes, p0 = 200, 120, 0.5
alpha = 0.05
phat = successes / n
se_null = np.sqrt(p0 * (1 - p0) / n)
z_stat = (phat - p0) / se_null
p_value = 2 * stats.norm.sf(abs(z_stat))

z_crit = stats.norm.ppf(1 - alpha / 2)
se_ci = np.sqrt(phat * (1 - phat) / n)
ci = (phat - z_crit * se_ci, phat + z_crit * se_ci)

print("Sample proportion:", phat)
print("SE under H0:", se_null)
print("z statistic:", z_stat)
print("p-value:", p_value)
print("95% Wald confidence interval:", ci)
```

## Question 5 — One-sample t-test

A t-test is commonly used to test a population mean when the population standard deviation is unknown and is estimated using the sample standard deviation. The t distribution accounts for this extra uncertainty. **Degrees of freedom** describe how many independent pieces of information remain after estimating parameters; for a one-sample t-test, df = n − 1.

Data: 48, 52, 50, 47, 53, 49, 51, 55, 46, 50. Test H₀: μ = 50 versus H₁: μ ≠ 50.

The sample mean is 50.1 and the sample standard deviation is approximately 2.7669. The standard error is \(s/\sqrt{10}\), so:

\[
t=\frac{50.1-50}{2.7669/\sqrt{10}}\approx0.1143
\]

Degrees of freedom = 9. The two-tailed critical value at α = 0.05 is approximately ±2.262. Since |t| < 2.262 (and p > 0.05), fail to reject H₀. The sample does not provide sufficient evidence that the population mean differs from 50.

```python
import numpy as np
from scipy import stats

data = np.array([48, 52, 50, 47, 53, 49, 51, 55, 46, 50], dtype=float)
mu0, alpha = 50, 0.05
n = len(data)
sample_mean = np.mean(data)
sample_sd = np.std(data, ddof=1)
t_stat = (sample_mean - mu0) / (sample_sd / np.sqrt(n))
df = n - 1
t_critical = stats.t.ppf(1 - alpha / 2, df)
result = stats.ttest_1samp(data, popmean=mu0)

print("Sample mean:", sample_mean)
print("Sample SD:", sample_sd)
print("t statistic (manual):", t_stat)
print("Degrees of freedom:", df)
print("Critical values:", -t_critical, t_critical)
print("SciPy t statistic and p-value:", result)
print("Decision:", "Reject H0" if result.pvalue < alpha else "Fail to reject H0")
```

## Question 6 — Independent two-sample t-test

Let the population means for methods A and B be μA and μB.

- H₀: μA = μB
- H₁: μA ≠ μB

The independent two-sample t-test compares the means of two separate, independent groups. A **pooled t-test** assumes equal population variances and combines the sample variance estimates. **Welch's t-test** does not assume equal variances and adjusts the degrees of freedom; in SciPy use `equal_var=False`.

Assumptions include independent observations, quantitative outcome, approximately normal scores within each group (especially important for small samples), and—only for the pooled test—equal population variances. Shapiro-Wilk can assess normality; Levene's test can assess equality of variances. These tests are diagnostics, not absolute proof of assumptions.

The sample means and sample variances are calculated below. Use Welch's test as a robust default if equal variances are not justified. A two-sided test at α = 0.05 assesses whether the mean scores differ.

```python
import numpy as np
from scipy import stats

A = np.array([78, 82, 85, 74, 80, 79, 83, 77], dtype=float)
B = np.array([72, 75, 70, 78, 74, 71, 76, 73], dtype=float)
alpha = 0.05

print("Method A mean:", np.mean(A), "sample variance:", np.var(A, ddof=1))
print("Method B mean:", np.mean(B), "sample variance:", np.var(B, ddof=1))

pooled = stats.ttest_ind(A, B, equal_var=True)
welch = stats.ttest_ind(A, B, equal_var=False)
print("\nPooled t-test:", pooled)
print("Welch t-test:", welch)

print("\nShapiro-Wilk normality checks:")
print("A:", stats.shapiro(A))
print("B:", stats.shapiro(B))
print("\nLevene's test for equal variances:", stats.levene(A, B))

print("Welch decision:", "Reject H0" if welch.pvalue < alpha else "Fail to reject H0")
```

## Question 7 — Paired t-test: before and after training

A paired t-test is appropriate because the before and after scores come from the **same eight students**. Each after-score is naturally paired with that student's before-score. An independent t-test would incorrectly treat the two sets as unrelated groups.

Define each difference as After − Before:

| Student | Before | After | Difference |
|---:|---:|---:|---:|
| 1 | 62 | 68 | 6 |
| 2 | 70 | 74 | 4 |
| 3 | 65 | 69 | 4 |
| 4 | 58 | 63 | 5 |
| 5 | 72 | 75 | 3 |
| 6 | 68 | 72 | 4 |
| 7 | 60 | 64 | 4 |
| 8 | 66 | 70 | 4 |

Mean difference = 4.25; sample standard deviation of differences ≈ 0.8864.

Hypotheses: H₀: mean difference = 0; H₁: mean difference ≠ 0. The paired t-test is equivalent to a one-sample t-test on the differences. The code verifies both approaches.

```python
import numpy as np
from scipy import stats

before = np.array([62, 70, 65, 58, 72, 68, 60, 66], dtype=float)
after = np.array([68, 74, 69, 63, 75, 72, 64, 70], dtype=float)
differences = after - before

paired_result = stats.ttest_rel(after, before)
difference_result = stats.ttest_1samp(differences, popmean=0)

print("Differences (after - before):", differences)
print("Mean difference:", np.mean(differences))
print("SD of differences:", np.std(differences, ddof=1))
print("Paired t-test:", paired_result)
print("One-sample t-test on differences:", difference_result)
print("Conclusion at 5%:", "Evidence of a change" if paired_result.pvalue < 0.05 else "Insufficient evidence of a change")
```

## Question 8 — Chi-square goodness-of-fit test

The chi-square goodness-of-fit test checks whether observed category counts are consistent with a specified probability distribution.

For a fair die rolled 60 times, the expected count for each face is:

\[
E_i=\frac{60}{6}=10
\]

Observed counts are 8, 9, 12, 11, 6, 14; expected counts are 10, 10, 10, 10, 10, 10.

Hypotheses:
- H₀: The die is fair (each face has probability 1/6).
- H₁: The die is not fair.

\[
\chi^2=\sum\frac{(O_i-E_i)^2}{E_i}
=\frac{4+1+4+1+16+16}{10}=4.2
\]

Degrees of freedom = k − 1 = 6 − 1 = 5. The critical value at α = 0.05 is approximately 11.070. The p-value is approximately 0.520. Since 4.2 < 11.070 and p > 0.05, fail to reject H₀. The observed counts do not provide sufficient evidence that the die is unfair.

```python
import numpy as np
from scipy import stats

observed = np.array([8, 9, 12, 11, 6, 14])
expected = np.full(6, 60 / 6)
chi_stat = np.sum((observed - expected) ** 2 / expected)
df = len(observed) - 1
critical = stats.chi2.ppf(0.95, df)
result = stats.chisquare(observed, f_exp=expected)

print("Expected counts:", expected)
print("Manual chi-square statistic:", chi_stat)
print("Degrees of freedom:", df)
print("Critical value:", critical)
print("SciPy result:", result)
print("Decision:", "Reject H0" if result.pvalue < 0.05 else "Fail to reject H0")
```

## Question 9 — Chi-square test of independence

The chi-square test of independence checks whether two categorical variables are associated.

Observed contingency table:

| Gender | Online | Store | Row total |
|---|---:|---:|---:|
| Male | 30 | 20 | 50 |
| Female | 15 | 35 | 50 |
| Column total | 45 | 55 | 100 |

Hypotheses:
- H₀: Gender and preferred shopping mode are independent.
- H₁: Gender and preferred shopping mode are associated.

Expected frequency for each cell:

\[
E_{ij}=\frac{(\text{row total})(\text{column total})}{\text{grand total}}
\]

Expected table:

| Gender | Online | Store |
|---|---:|---:|
| Male | 22.5 | 27.5 |
| Female | 22.5 | 27.5 |

Degrees of freedom = (rows − 1)(columns − 1) = 1.

The Pearson chi-square statistic is approximately 10.101 and the p-value is approximately 0.0015. Thus, there is evidence of an association between gender and preferred shopping mode at the 5% level.

**Yates' continuity correction** adjusts the Pearson chi-square statistic for a 2×2 table to improve the approximation to the continuous chi-square distribution, often making the test more conservative. SciPy's `chi2_contingency` applies it by default for a 2×2 table; set `correction=False` to obtain the uncorrected Pearson result.

```python
import numpy as np
from scipy import stats

observed_table = np.array([[30, 20],
                           [15, 35]])

row_totals = observed_table.sum(axis=1, keepdims=True)
col_totals = observed_table.sum(axis=0, keepdims=True)
grand_total = observed_table.sum()
expected_table = row_totals @ col_totals / grand_total

print("Expected frequencies:\n", expected_table)
print("Degrees of freedom:", (2 - 1) * (2 - 1))

# Default correction=True applies Yates' correction for a 2x2 table
with_yates = stats.chi2_contingency(observed_table, correction=True)
without_yates = stats.chi2_contingency(observed_table, correction=False)

print("\nWith Yates' correction:", with_yates)
print("Without Yates' correction:", without_yates)
print("Decision (uncorrected Pearson):",
      "Reject H0" if without_yates.pvalue < 0.05 else "Fail to reject H0")
```

## Question 10 — One-way ANOVA

**One-way analysis of variance (ANOVA)** tests whether the means of three or more independent groups are all equal.

- **Between-group variation (SSB):** Variation explained by differences between group means and the grand mean.
- **Within-group variation (SSW):** Variation of observations around their own group means.
- **Mean squares:** \(MSB=SSB/(k-1)\), and \(MSW=SSW/(N-k)\).
- **F statistic:** \(F=MSB/MSW\). A large F indicates that between-group variation is large relative to within-group variation.

Assumptions: independent observations, approximately normal outcomes within each group (or residuals), and equal population variances across groups.

Data:
- Group 1: 85, 88, 90, 84, 87; mean = 86.8
- Group 2: 78, 80, 82, 79, 81; mean = 80.0
- Group 3: 90, 92, 95, 91, 93; mean = 92.2

Grand mean = 86.3333.

\[
SSB=\sum n_i(\bar{x}_i-\bar{x})^2
=5[(86.8-86.3333)^2+(80-86.3333)^2+(92.2-86.3333)^2]
\approx374.9333
\]

\[
SSW=\sum_i\sum_j(x_{ij}-\bar{x}_i)^2=40.8
\]

With k = 3 groups and N = 15 observations:
- df between = 2; df within = 12
- \(MSB=374.9333/2\approx187.4667\)
- \(MSW=40.8/12=3.4\)
- \(F=187.4667/3.4\approx55.1373\)

At α = 0.05, compare this with the F critical value for (2, 12) degrees of freedom. A significant ANOVA indicates at least one group mean differs, but does not identify which groups differ. A post-hoc procedure such as **Tukey's HSD** is needed to make controlled pairwise comparisons while accounting for multiple testing.

```python
import numpy as np
from scipy import stats

g1 = np.array([85, 88, 90, 84, 87], dtype=float)
g2 = np.array([78, 80, 82, 79, 81], dtype=float)
g3 = np.array([90, 92, 95, 91, 93], dtype=float)
groups = [g1, g2, g3]

group_means = np.array([g.mean() for g in groups])
grand_mean = np.concatenate(groups).mean()
k = len(groups)
N = sum(len(g) for g in groups)

SSB = sum(len(g) * (g.mean() - grand_mean)**2 for g in groups)
SSW = sum(np.sum((g - g.mean())**2) for g in groups)
df_between = k - 1
df_within = N - k
MSB = SSB / df_between
MSW = SSW / df_within
F_manual = MSB / MSW
F_critical = stats.f.ppf(0.95, df_between, df_within)

scipy_result = stats.f_oneway(*groups)

print("Group means:", group_means)
print("Grand mean:", grand_mean)
print("SSB:", SSB)
print("SSW:", SSW)
print("MSB:", MSB)
print("MSW:", MSW)
print("Manual F statistic:", F_manual)
print("Degrees of freedom:", (df_between, df_within))
print("F critical at alpha=0.05:", F_critical)
print("SciPy f_oneway result:", scipy_result)
print("Decision:", "Reject H0" if scipy_result.pvalue < 0.05 else "Fail to reject H0")

# Optional post-hoc comparison (requires statsmodels, usually installed in Colab)
try:
    from statsmodels.stats.multicomp import pairwise_tukeyhsd
    all_scores = np.concatenate(groups)
    labels = (["Group 1"] * len(g1) +
              ["Group 2"] * len(g2) +
              ["Group 3"] * len(g3))
    print("\nTukey HSD post-hoc test:")
    print(pairwise_tukeyhsd(all_scores, labels, alpha=0.05))
except ImportError:
    print("Install statsmodels to run Tukey HSD: !pip install statsmodels")
```
